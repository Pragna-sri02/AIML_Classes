```python
#Employee class problem Create an Employee class with: Employee ID, Employee Name, and Salary. Create two employee objects and display their details.
class Employee:
    def __init__(self, id, name, salary):
        self.id = id
        self.name = name
        self.salary = salary

emp1 = Employee(101, "Rahul", 50000)
emp2 = Employee(102, "Anjali", 65000)

print("Employee ID:", emp1.id)
print("Employee Name:", emp1.name)
print("Salary:", emp1.salary)

print()

print("Employee ID:", emp2.id)
print("Employee Name:", emp2.name)
print("Salary:", emp2.salary)
```

```python
import numpy as np

# Generate marks for 500 students (0 to 100)
marks = np.random.randint(0, 101, 500)

# Count students in each range
between_60_80 = np.sum((marks >= 60) & (marks <= 80))
above_90 = np.sum(marks > 90)
below_50 = np.sum(marks < 50)

# Display results
print("Total Students:", len(marks))
print("Students scoring between 60 and 80:", between_60_80)
print("Students scoring above 90:", above_90)
print("Students scoring below 50:", below_50)
```

```python
import numpy as np
import pandas as pd
 
np.random.seed(42)
 
rows = 1000
 
df = pd.DataFrame({
"Student_ID": range(1, rows + 1),
"Age": np.random.randint(18, 26, rows),
"Study_Hours": np.random.randint(1, 11, rows),
"Attendance": np.random.randint(60, 101, rows),
"Assignments": np.random.randint(1, 11, rows),
"Previous_Marks": np.random.randint(40, 101, rows),
"Sleep_Hours": np.random.randint(4, 10, rows),
"Internet_Usage": np.random.randint(1, 9, rows),
"Marks": np.random.randint(35, 101, rows)
})
 
print(df.head())
 
df.to_csv("student_performance_1000.csv", index=False)
```

Object & Class

```python
class Student:

    def __init__(self, student_id, age, marks):
        self.student_id = student_id
        self.age = age
        self.marks = marks

    def display(self):
        print("Student ID:", self.student_id)
        print("Age:", self.age)
        print("Marks:", self.marks)

student = Student(df["Student_ID"][0], df["Age"][0], df["Marks"][0])

student.display()
```

```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
df=pd.read_excel(r'data.xlsx')
df
```

```python
#Q-Display employees with salary greater than ₹60,000.
salary = df[df["Salary"] > 60000]
print(salary)
```

```python
#Q-Find all female employees working in Bangalore.
female= df[(df["Gender"] == "f") & (df["JobCity"] == "Bangalore")]
print(female)
```

```python
#Q-Sort the dataset by English score in descending order.
dataset= df.sort_values(by="English", ascending=False)
print(dataset)
```

```python
#Q-Display the top 15 highest-paid employees.
employees = df.sort_values(by="Salary", ascending=False).head(15)
print(employees)
```

```python
#Q-Find the average salary for each city.
average_salary = df.groupby("JobCity")["Salary"].mean()
print(average_salary)
```

```python
#Q-Count the number of employees in each designation.
designation_count = df.groupby("Designation")["Designation"].count()
print(designation_count)
```

```python
#Q-Find the maximum and minimum salary for each gender
salary_for_gender= df.groupby("Gender")["Salary"].agg(["max", "min"])
print(salary_for_gender)
```

JobCity vs Salary

Bar Graph

```python
import matplotlib.pyplot as plt
avg_salary = df.groupby("JobCity")["Salary"].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
plt.bar(avg_salary.index, avg_salary.values, color=["skyblue","green","blue","red","yellow","purple","black","pink","grey","brown"])
plt.title("Average Salary by Job City")
plt.xlabel("Job City")
plt.ylabel("Average Salary")
plt.xticks(rotation=90)

plt.show()
```

line chart

```python
import matplotlib.pyplot as plt

avg_salary = df.groupby("JobCity")["Salary"].mean().sort_values(ascending=True).head(10)
plt.figure(figsize=(10,5))
plt.plot(avg_salary.index,
         avg_salary.values,
         marker="o",
         linestyle="-",
         color="blue")
plt.title("Average Salary by Job City")
plt.xlabel("Job City")
plt.ylabel("Average Salary")
plt.xticks(rotation=90)
plt.show()
```
